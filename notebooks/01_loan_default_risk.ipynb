```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option("display.max_columns", 150)
df = pd.read_csv("../data/application_train.csv")
print("Rows, columns:", df.shape)
df.head()
```

```python
default_rate = df["TARGET"].mean() * 100
print(f"Default rate: {default_rate:.2f}%")
print(df["TARGET"].value_counts())

sns.countplot(x="TARGET", data=df)
plt.title("Repaid (0) vs Defaulted (1)")
plt.show()
```

```python
missing = (df.isnull().mean() * 100).sort_values(ascending=False)
missing = missing[missing > 0]

print("Columns with missing values:", len(missing))
missing.head(15)
```

```python
df = df.copy()
df["AGE_YEARS"] = df["DAYS_BIRTH"] / -365

df["AGE_GROUP"] = pd.cut(df["AGE_YEARS"], bins=[20, 30, 40, 50, 60, 70])

age_default = df.groupby("AGE_GROUP", observed=True)["TARGET"].mean() * 100
print(age_default.round(2))

age_default.plot(kind="bar", color="steelblue")
plt.ylabel("Default rate (%)")
plt.title("Default rate by age group")
plt.show()
```

```python
missing_pct = df.isnull().mean() * 100
cols_to_drop = missing_pct[missing_pct > 50].index.tolist()

print("Columns dropped:", len(cols_to_drop))
df = df.drop(columns=cols_to_drop)
print("New shape:", df.shape)
```

```python
df["CREDIT_INCOME_RATIO"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
df["RATIO_BAND"] = pd.qcut(
    df["CREDIT_INCOME_RATIO"], 5,
    labels=["Lowest", "Low", "Mid", "High", "Highest"]
)
ratio_default = df.groupby("RATIO_BAND", observed=True)["TARGET"].mean() * 100
print(ratio_default.round(2))
ratio_default.plot(kind="bar", color="darkorange")
plt.ylabel("Default rate (%)")
plt.title("Default rate by loan-to-income band")
plt.show()
```

```python
edu_default = df.groupby("NAME_EDUCATION_TYPE")["TARGET"].mean() * 100
print(edu_default.sort_values().round(2))
edu_default.sort_values().plot(kind="barh", color="seagreen")
plt.xlabel("Default rate (%)")
plt.title("Default rate by education level")
plt.show()
```

```python
# How many customers are in each education group?
print(df["NAME_EDUCATION_TYPE"].value_counts())
# Monthly payment compared to income
df["ANNUITY_INCOME_RATIO"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
df["ANNUITY_BAND"] = pd.qcut(
    df["ANNUITY_INCOME_RATIO"], 5,
    labels=["Lowest", "Low", "Mid", "High", "Highest"]
)
annuity_default = df.groupby("ANNUITY_BAND", observed=True)["TARGET"].mean() * 100
print(annuity_default.round(2))

annuity_default.plot(kind="bar", color="purple")
plt.ylabel("Default rate (%)")
plt.title("Default rate by yearly payment-to-income band")
plt.show()
```

```python
if "EXT_SOURCE_1" not in df.columns:
    df["EXT_SOURCE_1"] = pd.read_csv(
        "../data/application_train.csv", usecols=["EXT_SOURCE_1"]
    )["EXT_SOURCE_1"]

ext_cols = ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]
# Does a higher external score go with a lower default rate?
print(df[ext_cols + ["TARGET"]].corr()["TARGET"].round(3))
df["EXT2_BAND"] = pd.qcut(df["EXT_SOURCE_2"], 5,
                          labels=["Lowest", "Low", "Mid", "High", "Highest"])
ext2_default = df.groupby("EXT2_BAND", observed=True)["TARGET"].mean() * 100
print(ext2_default.round(2))

ext2_default.plot(kind="bar", color="crimson")
plt.ylabel("Default rate (%)")
plt.title("Default rate by external credit score band")
plt.show()
```

```python
ext_cols = ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]

# 365243 days is a placeholder, so I flag it and blank it out
df["NOT_EMPLOYED_FLAG"] = (df["DAYS_EMPLOYED"] == 365243).astype(int)
df["DAYS_EMPLOYED"] = df["DAYS_EMPLOYED"].replace(365243, np.nan)
df["EMPLOYED_YEARS"] = df["DAYS_EMPLOYED"] / -365

df["CREDIT_TERM"] = df["AMT_ANNUITY"] / df["AMT_CREDIT"]
df["EXT_SOURCE_MEAN"] = df[ext_cols].mean(axis=1)

# dropping the id and the chart helper columns
drop_cols = ["SK_ID_CURR", "AGE_GROUP", "RATIO_BAND", "ANNUITY_BAND", "EXT2_BAND"]
df_model = df.drop(columns=drop_cols)

print(df_model.shape)
print(df_model["NOT_EMPLOYED_FLAG"].sum())
```

```python
from sklearn.model_selection import train_test_split

X = df_model.drop(columns=["TARGET"])
y = df_model["TARGET"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
print(round(y_train.mean() * 100, 2), round(y_test.mean() * 100, 2))
```

```python
X_train = pd.get_dummies(X_train, dtype=int)
X_test = pd.get_dummies(X_test, dtype=int)
X_train, X_test = X_train.align(X_test, join="left", axis=1, fill_value=0)

# medians come from train only so the test stays fair
medians = X_train.median()
X_train_filled = X_train.fillna(medians)
X_test_filled = X_test.fillna(medians)

print(X_train_filled.shape, X_test_filled.shape)
print(X_train_filled.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, recall_score, precision_score

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train_filled)
X_test_s = scaler.transform(X_test_filled)

log_model = LogisticRegression(class_weight="balanced", max_iter=1000)
log_model.fit(X_train_s, y_train)

log_probs = log_model.predict_proba(X_test_s)[:, 1]
log_preds = (log_probs >= 0.5).astype(int)

print("AUC:", round(roc_auc_score(y_test, log_probs), 3))
print("Recall:", round(recall_score(y_test, log_preds), 3))
print("Precision:", round(precision_score(y_test, log_preds), 3))
```

```python
import re
from xgboost import XGBClassifier

# xgboost doesn't like some symbols in column names
X_train_x = X_train.copy()
X_test_x = X_test.copy()
X_train_x.columns = [re.sub(r"[^A-Za-z0-9_]", "_", c) for c in X_train_x.columns]
X_test_x.columns = X_train_x.columns

ratio = (y_train == 0).sum() / (y_train == 1).sum()

xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    scale_pos_weight=ratio,
    tree_method="hist",
    random_state=42,
)
xgb_model.fit(X_train_x, y_train)

xgb_probs = xgb_model.predict_proba(X_test_x)[:, 1]
xgb_preds = (xgb_probs >= 0.5).astype(int)

print("AUC:", round(roc_auc_score(y_test, xgb_probs), 3))
print("Recall:", round(recall_score(y_test, xgb_preds), 3))
print("Precision:", round(precision_score(y_test, xgb_preds), 3))
```

```python
from sklearn.metrics import roc_curve

fpr1, tpr1, _ = roc_curve(y_test, log_probs)
fpr2, tpr2, _ = roc_curve(y_test, xgb_probs)

plt.figure(figsize=(6, 5))
plt.plot(fpr1, tpr1, label="Logistic regression")
plt.plot(fpr2, tpr2, label="XGBoost")
plt.plot([0, 1], [0, 1], "--", color="grey")
plt.xlabel("Good customers wrongly flagged")
plt.ylabel("Defaulters caught")
plt.legend()
plt.title("ROC curve")
plt.show()

imp = pd.Series(xgb_model.feature_importances_, index=X_train_x.columns)
imp.sort_values().tail(10).plot(kind="barh", color="teal")
plt.title("Top 10 features in XGBoost")
plt.show()
```

```python
no_gender = [c for c in X_train_x.columns if not c.startswith("CODE_GENDER")]

xgb_fair = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    scale_pos_weight=ratio,
    tree_method="hist",
    random_state=42,
)
xgb_fair.fit(X_train_x[no_gender], y_train)

fair_probs = xgb_fair.predict_proba(X_test_x[no_gender])[:, 1]
print("AUC without gender:", round(roc_auc_score(y_test, fair_probs), 3))
```

```python
margin = 0.15
loss = 0.60

test = pd.DataFrame({
    "prob": fair_probs,
    "default": y_test.values,
    "credit": df_model.loc[X_test.index, "AMT_CREDIT"].values,
})
test = test.sort_values("prob").reset_index(drop=True)

test["profit"] = np.where(test["default"] == 0,
                          test["credit"] * margin,
                          -test["credit"] * loss)
test["cum_profit"] = test["profit"].cumsum()
test["share_approved"] = (test.index + 1) / len(test)

best = test["cum_profit"].idxmax()
print("Profit if I approve everyone (millions):", round(test["profit"].sum() / 1e6, 1))
print("Profit at best cut-off (millions):", round(test["cum_profit"].max() / 1e6, 1))
print("Share of customers approved:", round(test.loc[best, "share_approved"] * 100, 1), "%")
print("Cut-off score:", round(test.loc[best, "prob"], 3))

approved = test.loc[:best]
rejected = test.loc[best + 1:]
print("Default rate approved:", round(approved["default"].mean() * 100, 2), "%")
print("Default rate rejected:", round(rejected["default"].mean() * 100, 2), "%")

plt.plot(test["share_approved"] * 100, test["cum_profit"] / 1e6)
plt.axvline(test.loc[best, "share_approved"] * 100, color="red", linestyle="--")
plt.xlabel("% of customers approved (safest first)")
plt.ylabel("Profit (millions)")
plt.title("Profit by approval cut-off")
plt.show()
```

```python
cut = (best + 1) / len(test)

test["band"] = pd.cut(test["share_approved"],
                      bins=[0, 0.6, 0.8, cut, 1.0],
                      labels=["Low", "Medium", "High", "Reject"])

summary = test.groupby("band", observed=True).agg(
    customers=("default", "size"),
    default_rate=("default", "mean"),
    avg_loan=("credit", "mean"),
    profit=("profit", "sum"),
)
summary["default_rate"] = (summary["default_rate"] * 100).round(2)
summary["avg_loan"] = summary["avg_loan"].round(0)
summary["profit"] = (summary["profit"] / 1e6).round(1)
print(summary)

# reject band profit is negative, that's the loss I avoid by not lending
summary.to_csv("../outputs/risk_bands.csv")
test.to_csv("../outputs/scored_test_customers.csv", index=False)
```

```python
def best_cut(margin, loss):
    p = np.where(test["default"] == 0, test["credit"] * margin, -test["credit"] * loss)
    cum = p.cumsum()
    i = cum.argmax()
    return round((i + 1) / len(test) * 100, 1), round(cum.max() / 1e6, 1), round(p.sum() / 1e6, 1)

for l in [0.4, 0.6, 0.8, 1.0]:
    share, best_p, all_p = best_cut(0.15, l)
    print("loss", l, "| approve", share, "% | best", best_p, "| everyone", all_p)
```

```python
rejected_mask = test.index > best
caught = test.loc[rejected_mask, "default"].sum() / test["default"].sum()
print("Defaulters caught by the cut-off:", round(caught * 100, 1), "%")

top20 = test.index >= int(len(test) * 0.8)
share20 = test.loc[top20, "default"].sum() / test["default"].sum()
print("Defaulters in the riskiest 20%:", round(share20 * 100, 1), "%")
```
